## 🎯 Learning Objectives

- Understand tokenization
- Convert text → tokens
- Use a fixed embedding table (3 dimensions)
- Build context windows (size = 2)
- Construct a 3D tensor `(3, 2, 3)` ready for a mini GPT

---

## 🧩 Sentence and Tokenization

Sentence:
```
I love large language models
```


````python
sentence = "I love large language models"
tokens = sentence.split()
tokens
````

---

## ✂️ Context Length = 2

We slide a 2-token window across the sentence.

```
[I, love] → predict "large"
[love, large] → predict "language"
[large, language] → predict "models"
```

There are **3 context windows**.

---

## 🧠 Fixed 3‑Dimensional Embeddings

These are **not random** and **not PCA** – they are fixed values for teaching.


````python
import numpy as np
import torch

emb3d = {
    "I":        np.array([0.2, 0.1, -0.3]),
    "love":     np.array([0.8, -0.2, 0.5]),
    "large":    np.array([0.1, 0.6, -0.1]),
    "language": np.array([0.4, 0.9, 0.2]),
    "models":   np.array([-0.2, 0.3, 0.7]),
}

emb3d
````

---

## 🧮 Build Context Windows


````python
context_len = 2
sequences = []

for i in range(len(tokens) - context_len):
    w1 = tokens[i]
    w2 = tokens[i+1]
    seq = np.vstack([emb3d[w1], emb3d[w2]])  # shape (2, 3)
    sequences.append(seq)

len(sequences)
````

---

## 📦 Final 3D Tensor


````python
X = torch.tensor(np.stack(sequences), dtype=torch.float32)
X, X.shape
````

Expected shape:
```
(3, 2, 3)
```

---

## 🎨 Visualizing the Embeddings


````python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(7,6))
ax = fig.add_subplot(111, projection='3d')

colors = ['red','green','blue']

for i, seq in enumerate(X):
    xs, ys, zs = seq[:,0], seq[:,1], seq[:,2]
    ax.plot(xs, ys, zs, marker='o', color=colors[i], label=f"Seq {i+1}")

ax.set_xlabel('Dim 1')
ax.set_ylabel('Dim 2')
ax.set_zlabel('Dim 3')
ax.set_title('3D Embedding Visualization (Context=2, Dim=3)')
ax.legend()
plt.show()
````

## 📊 Tokenization & Data Scale Comparison

| Feature | **Tiny Example** | **ChatGPT (GPT‑4/4o)** | **Gemini 1.5** | **Grok‑2** |
|--------|-------------------|------------------------|----------------|------------|
| Tokenization | Word-split | BPE | SentencePiece | SP/BPE hybrid |
| Vocabulary size | ~10 words | ~100k–200k | ~250k | ~200k |
| Context length | **2** | 8k–128k | **1,000,000** | 128k |
| Embedding dim | **3** | 3k–12k | 2k–16k | 4k–8k |
