AdaBoost constructs an ensemble by training a sequence of simple rules (**decision stumps**), where each successive round focuses on the errors of the previous ones.

```
Round 1                     Round 2                     Round 3
┌──────────────┐            ┌──────────────┐            ┌──────────────┐
│  All points  │ ──errors──►│ Hard points  │ ──errors──►│ Hardest pts  │
│  equal size  │    grow    │ grow larger  │    grow    │ dominate cut │
│   ●  ●  ●    │            │   ●  ⬤  ●    │            │   ●  ██  ●   │
└──────────────┘            └──────────────┘            └──────────────┘
       │                           │                           │
       ▼                           ▼                           ▼
Stump 1 (Vote: α₁)          Stump 2 (Vote: α₂)          Stump 3 (Vote: α₃)
       └───────────────────────────┬───────────────────────────┘
                                   ▼
              Ensemble: H(x) = sign( α₁h₁ + α₂h₂ + α₃h₃ )
```

---

### 1. The 3-Step Iteration Loop

| Step | Action | Visual in Widget |
| :--- | :--- | :--- |
| **1. Split** | Fits a single axis-aligned cut minimizing weighted error $\epsilon_m$. | **Black line** on left plot |
| **2. Grade** | Computes voting power: $\alpha_m = \frac{1}{2}\ln\left(\frac{1-\epsilon_m}{\epsilon_m}\right)$. | **Green metric card** |
| **3. Reweight** | Misclassified points **expand**; correctly classified points **shrink**. | **Yellow rings** $\rightarrow$ larger bubbles next round |

---

### 2. Error Rate vs. Voting Weight ($\alpha_m$)

$$\alpha_m = \frac{1}{2} \ln\left(\frac{1 - \epsilon_m}{\epsilon_m}\right)$$

| Performance | Weighted Error ($\epsilon_m$) | Voting Power ($\alpha_m$) | Ensemble Impact |
| :---: | :---: | :---: | :--- |
| **Near-Perfect** | $0.05$ | **$+1.47$** | Dominates the ensemble decision |
| **Slight Edge** | $0.35$ | **$+0.31$** | Provides a tie-breaking vote |
| **Coin Flip** | $0.50$ | **$0.00$** | **Ignored** (casts no vote) |
| **Worse Than Random** | $0.80$ | **$-0.69$** | Prediction inverted by negative sign |

---

### 3. Reading the Dashboard

```
   ┌───────────────────────────────┐     ┌───────────────────────────────┐
   │       LEFT PLOT: Stump m      │     │      RIGHT PLOT: Ensemble     │
   ├───────────────────────────────┤     ├───────────────────────────────┤
   │ • Single straight cut (black) │     │ • Combined voting surface     │
   │ • Circle size = current weight│     │ • Smooth, curved boundary     │
   │ • Yellow ring = stump error   │     │ • Overall train accuracy      │
   └───────────────────────────────┘     └───────────────────────────────┘
```

> **Key Rule:** A single stump cannot bend. Curved contours on the right emerge strictly because stumps with different cut angles combine through weighted addition.

````python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import HTML, display
from matplotlib.colors import ListedColormap
from sklearn.datasets import make_blobs, make_circles, make_moons
from sklearn.tree import DecisionTreeClassifier

# -----------------------------------------------------------------------------
# 1. Transparent AdaBoost Engine (Step-by-Step State Tracker)
# -----------------------------------------------------------------------------
class AdaBoostTracker:
    """Trains AdaBoost.M1 and stores intermediate states for educational inspection."""

    def __init__(self, X, y, max_rounds=20):
        self.X = X
        # Map labels to {-1, +1} for classic AdaBoost arithmetic
        self.y = np.where(y == 0, -1, 1)
        self.N = len(X)
        self.max_rounds = max_rounds

        self.stumps = []
        self.alphas = []
        self.weighted_errors = []
        # weights_history[m] stores sample weights at the start of round m
        self.weights_history = [np.ones(self.N) / self.N]
        self.stump_preds = []
        self.cumulative_train_acc = []

        self._fit()

    def _fit(self):
        w = np.ones(self.N) / self.N
        cum_scores = np.zeros(self.N)

        for m in range(self.max_rounds):
            # 1. Fit weak learner (decision stump) using current sample weights
            stump = DecisionTreeClassifier(max_depth=1, random_state=42)
            stump.fit(self.X, self.y, sample_weight=w)
            pred = stump.predict(self.X)

            # 2. Weighted classification error
            misclassified = pred != self.y
            epsilon = np.sum(w[misclassified]) / np.sum(w)
            epsilon = np.clip(epsilon, 1e-10, 1 - 1e-10)

            # 3. Estimator weight (alpha)
            alpha = 0.5 * np.log((1.0 - epsilon) / epsilon)

            # 4. Update sample weights: w <- w * exp(-alpha * y * pred)
            w = w * np.exp(-alpha * self.y * pred)
            w = w / np.sum(w)  # Re-normalize

            # Record round variables
            self.stumps.append(stump)
            self.alphas.append(alpha)
            self.weighted_errors.append(epsilon)
            self.weights_history.append(w.copy())
            self.stump_preds.append(pred)

            # Running ensemble accuracy
            cum_scores += alpha * pred
            cum_acc = np.mean(np.sign(cum_scores) == self.y)
            self.cumulative_train_acc.append(cum_acc)


# -----------------------------------------------------------------------------
# 2. Data Synthesis Function
# -----------------------------------------------------------------------------
def get_dataset(name, n_samples=50, noise=0.2, seed=42):
    np.random.seed(seed)
    if name == "Moons":
        X, y = make_moons(n_samples=n_samples, noise=noise, random_state=seed)
    elif name == "Circles":
        X, y = make_circles(
            n_samples=n_samples, noise=noise, factor=0.5, random_state=seed
        )
    else:  # Two clusters with overlapping outliers
        X, y = make_blobs(
            n_samples=n_samples,
            centers=[[-1.2, -1.0], [1.2, 1.0]],
            cluster_std=1.0 + noise * 1.5,
            random_state=seed,
        )
    return X, y


# -----------------------------------------------------------------------------
# 3. Interactive UI & Plotting Dashboard
# -----------------------------------------------------------------------------
# Global domain bounds
X_LIM = (-3.2, 3.2)
Y_LIM = (-3.2, 3.2)
CMAP_BG = ListedColormap(["#ffebee", "#e3f2fd"])

# Widget Controls
dataset_dd = widgets.Dropdown(
    options=["Moons", "Circles", "Overlapping Blobs"],
    value="Moons",
    description="Dataset:",
    layout=widgets.Layout(width="220px"),
)

samples_slider = widgets.IntSlider(
    value=50,
    min=20,
    max=100,
    step=10,
    description="Samples (N):",
    layout=widgets.Layout(width="240px"),
)

noise_slider = widgets.FloatSlider(
    value=0.20,
    min=0.05,
    max=0.50,
    step=0.05,
    description="Noise:",
    layout=widgets.Layout(width="220px"),
)

round_slider = widgets.IntSlider(
    value=1,
    min=1,
    max=15,
    step=1,
    description="Round (m):",
    layout=widgets.Layout(width="260px"),
)

btn_step_prev = widgets.Button(
    description="◀ Prev", layout=widgets.Layout(width="80px")
)
btn_step_next = widgets.Button(
    description="Next ▶",
    button_style="primary",
    layout=widgets.Layout(width="80px"),
)
btn_new_seed = widgets.Button(
    description="Resample", icon="refresh", layout=widgets.Layout(width="110px")
)

chk_weight_sizes = widgets.Checkbox(
    value=True,
    description="Scale points by weight $w_i$",
    indent=False,
    layout=widgets.Layout(width="230px"),
)
chk_highlight_errors = widgets.Checkbox(
    value=True,
    description="Highlight round misclassifications",
    indent=False,
    layout=widgets.Layout(width="260px"),
)

output_plot = widgets.Output()
output_stats = widgets.Output()

# Application state
state = {"seed": 42, "model": None, "grid_X": None}

# Precompute grid mesh for boundaries
gx = np.linspace(X_LIM[0], X_LIM[1], 150)
gy = np.linspace(Y_LIM[0], Y_LIM[1], 150)
XX, YY = np.meshgrid(gx, gy)
grid_points = np.c_[XX.ravel(), YY.ravel()]


def retrain_model():
    X, y = get_dataset(
        dataset_dd.value,
        n_samples=samples_slider.value,
        noise=noise_slider.value,
        seed=state["seed"],
    )
    state["model"] = AdaBoostTracker(
        X, y, max_rounds=round_slider.max
    )


def render():
    m_idx = round_slider.value - 1  # 0-indexed round
    model = state["model"]

    current_stump = model.stumps[m_idx]
    current_alpha = model.alphas[m_idx]
    current_eps = model.weighted_errors[m_idx]
    current_weights = model.weights_history[m_idx]
    current_pred = model.stump_preds[m_idx]
    misclassified = current_pred != model.y

    # Compute cumulative ensemble prediction up to round m
    cum_grid = np.zeros(grid_points.shape[0])
    for k in range(m_idx + 1):
        cum_grid += model.alphas[k] * model.stumps[k].predict(grid_points)
    cum_decision_mesh = np.sign(cum_grid).reshape(XX.shape)

    # Compute weak learner boundary on grid
    stump_decision_mesh = current_stump.predict(grid_points).reshape(XX.shape)

    with output_plot:
        output_plot.clear_output(wait=True)
        fig, axes = plt.subplots(1, 2, figsize=(14, 5.8), sharey=True)

        for ax in axes:
            ax.set_xlim(X_LIM)
            ax.set_ylim(Y_LIM)
            ax.grid(True, linestyle=":", alpha=0.5)

        # Marker sizes: scale base size (40) by normalized sample weight
        if chk_weight_sizes.value:
            # Multiply by N so an average-weighted sample has size ~45
            sizes = 45 * (current_weights * model.N)
            sizes = np.clip(sizes, 10, 450)
        else:
            sizes = np.full(model.N, 45)

        color_map = np.where(model.y == 1, "#1976d2", "#d32f2f")

        # -------------------------------------------------------------
        # Left Plot: Current Decision Stump m
        # -------------------------------------------------------------
        axes[0].contourf(
            XX,
            YY,
            stump_decision_mesh,
            levels=[-1.5, 0, 1.5],
            cmap=CMAP_BG,
            alpha=0.6,
        )
        axes[0].contour(
            XX,
            YY,
            stump_decision_mesh,
            levels=[0],
            colors="#212121",
            linewidths=2.0,
        )

        axes[0].scatter(
            model.X[:, 0],
            model.X[:, 1],
            s=sizes,
            c=color_map,
            edgecolors="black",
            linewidth=0.8,
            zorder=4,
        )

        if chk_highlight_errors.value and np.any(misclassified):
            axes[0].scatter(
                model.X[misclassified, 0],
                model.X[misclassified, 1],
                s=sizes[misclassified] + 80,
                facecolors="none",
                edgecolors="#ffd600",
                linewidth=2.5,
                label="Wrong by Stump",
                zorder=5,
            )
            axes[0].legend(loc="upper right", framealpha=0.9)

        axes[0].set_title(
            f"Round {m_idx + 1} Weak Learner $h_{{{m_idx + 1}}}(x)$ (Decision"
            " Stump)\nMarker size reflects training weight $w_i$",
            fontsize=11,
            fontweight="bold",
        )
        axes[0].set_xlabel("$x_1$")
        axes[0].set_ylabel("$x_2$")

        # -------------------------------------------------------------
        # Right Plot: Cumulative Ensemble H_m(x)
        # -------------------------------------------------------------
        axes[1].contourf(
            XX,
            YY,
            cum_decision_mesh,
            levels=[-1.5, 0, 1.5],
            cmap=CMAP_BG,
            alpha=0.6,
        )
        axes[1].contour(
            XX,
            YY,
            cum_decision_mesh,
            levels=[0],
            colors="#1b5e20",
            linewidths=2.4,
        )

        axes[1].scatter(
            model.X[:, 0],
            model.X[:, 1],
            s=45,
            c=color_map,
            edgecolors="black",
            linewidth=0.8,
            zorder=4,
        )

        axes[1].set_title(
            f"Ensemble $H_{{{m_idx + 1}}}(x) = "
            r"\mathrm{sign}\left(\sum_{k=1}^{"
            f"{m_idx + 1}"
            r"}\alpha_k h_k(x)\right)$"
            f"\nTrain Accuracy: {model.cumulative_train_acc[m_idx]:.1%}",
            fontsize=11,
            fontweight="bold",
        )
        axes[1].set_xlabel("$x_1$")

        plt.tight_layout()
        plt.show()

    # Dynamic Math & Metrics readout
    with output_stats:
        output_stats.clear_output(wait=True)
        html_card = f"""
        <div style="background:#f8f9fa; border:1px solid #dee2e6; border-radius:6px; padding:12px 18px; margin-top:8px; font-family:sans-serif; font-size:13px; line-height:1.6;">
            <div style="display:flex; justify-content:space-between; flex-wrap:wrap; gap:12px;">
                <div>
                    <b>Stump Weighted Error ($\epsilon_{{{m_idx+1}}}$):</b> 
                    <span style="color:#c62828; font-weight:bold;">{current_eps:.4f}</span>
                    <span style="color:#757575;">(Misclassified: {np.sum(misclassified)} / {model.N} points)</span>
                </div>
                <div>
                    <b>Stump Weight Vote ($\alpha_{{{m_idx+1}}}$):</b> 
                    <span style="color:#2e7d32; font-weight:bold;">{current_alpha:.4f}</span>
                    <span style="color:#757575;">[Formula: $\\frac{{1}}{{2}}\\ln\\frac{{1-\\epsilon}}{{\\epsilon}}$]</span>
                </div>
                <div>
                    <b>Overall Ensemble Accuracy:</b> 
                    <span style="color:#1565c0; font-weight:bold;">{model.cumulative_train_acc[m_idx]:.1%}</span>
                </div>
            </div>
            <div style="margin-top:8px; color:#424242;">
                💡 <b>Teaching Insight:</b> Notice how the misclassified points inside the yellow rings receive higher weights $w_i$, making their bubbles larger. In Round {min(m_idx+2, round_slider.max)}, the next stump is heavily penalized if it does not separate them.
            </div>
        </div>
        """
        display(HTML(html_card))


# -----------------------------------------------------------------------------
# 4. Callback Connections
# -----------------------------------------------------------------------------
def on_dataset_or_noise_change(change):
    retrain_model()
    render()


def on_round_change(change):
    render()


def on_step_next_clicked(b):
    if round_slider.value < round_slider.max:
        round_slider.value += 1


def on_step_prev_clicked(b):
    if round_slider.value > round_slider.min:
        round_slider.value -= 1


def on_resample_clicked(b):
    state["seed"] = np.random.randint(1, 10000)
    retrain_model()
    render()


dataset_dd.observe(on_dataset_or_noise_change, names="value")
samples_slider.observe(on_dataset_or_noise_change, names="value")
noise_slider.observe(on_dataset_or_noise_change, names="value")
round_slider.observe(on_round_change, names="value")
chk_weight_sizes.observe(lambda c: render(), names="value")
chk_highlight_errors.observe(lambda c: render(), names="value")

btn_step_next.on_click(on_step_next_clicked)
btn_step_prev.on_click(on_step_prev_clicked)
btn_new_seed.on_click(on_resample_clicked)

# -----------------------------------------------------------------------------
# 5. Dashboard Layout Assembly
# -----------------------------------------------------------------------------
row1 = widgets.HBox(
    [dataset_dd, samples_slider, noise_slider, btn_new_seed],
    layout=widgets.Layout(align_items="center", gap="12px"),
)
row2 = widgets.HBox(
    [btn_step_prev, btn_step_next, round_slider, chk_weight_sizes, chk_highlight_errors],
    layout=widgets.Layout(align_items="center", gap="12px", margin="8px 0"),
)

display(widgets.VBox([row1, row2, output_stats, output_plot]))

# Initial setup execution
retrain_model()
render()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_gaussian_quantiles

# 1. Generate a non-linearly separable toy dataset
np.random.seed(42)
n_samples = 5

# Create two concentric clusters for class +1 and class -1
X1, y1 = make_gaussian_quantiles(cov=2.0, n_samples=n_samples, n_features=2, n_classes=2, shuffle=True, random_state=1)
X2, y2 = make_gaussian_quantiles(mean=(3, 3), cov=1.5, n_samples=n_samples, n_features=2, n_classes=2, shuffle=True, random_state=2)

X = np.vstack((X1, X2))
y = np.hstack((y1, 1 - y2))
y = np.where(y == 0, -1, 1)  # Labels must be in {-1, +1} for AdaBoost

# 2. Decision Stump Classifier (Weak Learner)
class DecisionStump:
    def __init__(self):
        self.polarity = 1
        self.feature_idx = None
        self.threshold = None
        self.alpha = None

    def predict(self, X):
        n_samples = X.shape[0]
        X_column = X[:, self.feature_idx]
        predictions = np.ones(n_samples)
        
        if self.polarity == 1:
            predictions[X_column < self.threshold] = -1
        else:
            predictions[X_column > self.threshold] = -1
        return predictions

# 3. Custom AdaBoost Implementation
class AdaBoostScratch:
    def __init__(self, n_estimators=5):
        self.n_estimators = n_estimators
        self.clfs = []
        self.weights_history = []  # Stores weights after each round

    def fit(self, X, y):
        n_samples, n_features = X.shape
        w = np.full(n_samples, (1 / n_samples))

        for _ in range(self.n_estimators):
            clf = DecisionStump()
            min_error = float('inf')

            # Exhaustive search for best feature, threshold, and polarity
            for feature_i in range(n_features):
                X_column = X[:, feature_i]
                thresholds = np.unique(X_column)

                for threshold in thresholds:
                    for polarity in [1, -1]:
                        predictions = np.ones(n_samples)
                        if polarity == 1:
                            predictions[X_column < threshold] = -1
                        else:
                            predictions[X_column > threshold] = -1

                        # Calculate weighted classification error
                        misclassified = w[y != predictions]
                        error = np.sum(misclassified)

                        if error < min_error:
                            min_error = error
                            clf.polarity = polarity
                            clf.threshold = threshold
                            clf.feature_idx = feature_i

            # Numerical stability eps
            eps = 1e-10
            min_error = np.clip(min_error, eps, 1.0 - eps)

            # Compute stump weight (alpha)
            clf.alpha = 0.5 * np.log((1.0 - min_error) / min_error)

            # Update sample weights
            predictions = clf.predict(X)
            w *= np.exp(-clf.alpha * y * predictions)
            w /= np.sum(w)  # Re-normalize

            # Save state
            self.clfs.append(clf)
            self.weights_history.append(w.copy())

    def predict_up_to_round(self, X, round_idx):
        """Ensemble decision up to round_idx (inclusive)"""
        clf_preds = [clf.alpha * clf.predict(X) for clf in self.clfs[:round_idx + 1]]
        return np.sign(np.sum(clf_preds, axis=0))

# 4. Train AdaBoost
n_rounds = 4
model = AdaBoostScratch(n_estimators=n_rounds)
model.fit(X, y)

# 5. Visualization
# Generate mesh grid for plotting decision surfaces
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300), np.linspace(y_min, y_max, 300))
grid_points = np.c_[xx.ravel(), yy.ravel()]

fig, axes = plt.subplots(n_rounds, 2, figsize=(14, 4 * n_rounds))
plt.subplots_adjust(hspace=0.35, wspace=0.25)

for r in range(n_rounds):
    clf = model.clfs[r]
    weights = model.weights_history[r]
    
    # Scale point sizes for visual clarity (proportional to weights)
    point_sizes = weights * len(X) * 45

    # ---------------- Left Plot: Weak Learner Boundary & Point Weights ----------------
    ax_left = axes[r, 0]
    Z_stump = clf.predict(grid_points).reshape(xx.shape)
    
    ax_left.contourf(xx, yy, Z_stump, levels=[-1, 0, 1], alpha=0.25, cmap=plt.cm.coolwarm)
    ax_left.contour(xx, yy, Z_stump, levels=[0], colors='black', linewidths=1.5, linestyles='--')
    
    # Scatter points with size mapped to weights
    scatter_pos = ax_left.scatter(X[y == 1, 0], X[y == 1, 1], s=point_sizes[y == 1],
                                 c='red', edgecolors='k', label='Class +1', alpha=0.8)
    scatter_neg = ax_left.scatter(X[y == -1, 0], X[y == -1, 1], s=point_sizes[y == -1],
                                 c='blue', edgecolors='k', label='Class -1', alpha=0.8)
    
    ax_left.set_title(f"Round {r+1}: Weak Learner (Stump) Boundary\nStump Weight ($\\alpha$): {clf.alpha:.3f}")
    ax_left.set_xlabel("$X_1$")
    ax_left.set_ylabel("$X_2$")
    if r == 0:
        ax_left.legend(loc='upper right')

    # ---------------- Right Plot: Cumulative Ensemble Boundary ----------------
    ax_right = axes[r, 1]
    Z_ensemble = model.predict_up_to_round(grid_points, r).reshape(xx.shape)
    
    ax_right.contourf(xx, yy, Z_ensemble, levels=[-1, 0, 1], alpha=0.35, cmap=plt.cm.coolwarm)
    ax_right.contour(xx, yy, Z_ensemble, levels=[0], colors='black', linewidths=2)
    
    # Static point sizes on ensemble map to focus on the overall boundary
    ax_right.scatter(X[y == 1, 0], X[y == 1, 1], s=35, c='red', edgecolors='k', alpha=0.8)
    ax_right.scatter(X[y == -1, 0], X[y == -1, 1], s=35, c='blue', edgecolors='k', alpha=0.8)
    
    ax_right.set_title(f"Cumulative Ensemble Boundary (Rounds 1 to {r+1})")
    ax_right.set_xlabel("$X_1$")
    ax_right.set_ylabel("$X_2$")

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from sklearn.datasets import make_gaussian_quantiles
from sklearn.tree import DecisionTreeClassifier

# ==============================================================================
# 1. Configuration & Dataset Generation
# ==============================================================================
N_SAMPLES = 5       # Adjust number of data points here
N_ROUNDS = 4         # AdaBoost iterations (designed for a 2x2 grid on the left)
RANDOM_STATE = 42

# Generate a 2D synthetic dataset with two overlapping clusters
X1, y1 = make_gaussian_quantiles(
    cov=1.8, n_samples=int(N_SAMPLES * 0.5), n_features=2, 
    n_classes=1, random_state=RANDOM_STATE
)
X2, y2 = make_gaussian_quantiles(
    mean=(2.0, 2.0), cov=1.4, n_samples=int(N_SAMPLES * 0.5), 
    n_features=2, n_classes=1, random_state=RANDOM_STATE + 1
)

X = np.vstack([X1, X2])
# Encode targets into {-1, +1} for AdaBoost math
y = np.hstack([np.ones(len(X1)), -np.ones(len(X2))])

# ==============================================================================
# 2. AdaBoost Step-by-Step Training
# ==============================================================================
n_samples = X.shape[0]

# Step A: Initialize uniform sample weights
w = np.full(n_samples, 1.0 / n_samples)

stumps = []
alphas = []
errors = []
weights_history = [w.copy()]

for t in range(N_ROUNDS):
    # Step B: Train weak classifier (depth-1 decision stump) using sample weights
    stump = DecisionTreeClassifier(max_depth=1, random_state=RANDOM_STATE)
    stump.fit(X, y, sample_weight=w)
    pred = stump.predict(X)
    
    # Step C: Compute weighted classification error
    misclassified = (pred != y)
    err = np.sum(w[misclassified]) / np.sum(w)
    err = np.clip(err, 1e-10, 1.0 - 1e-10)  # Numerical stability against division by zero
    
    # Step D: Compute weak learner voting power (alpha)
    alpha = 0.5 * np.log((1.0 - err) / err)
    
    # Step E: Update sample weights (increase for errors, decrease for correct)
    w = w * np.exp(-alpha * y * pred)
    w /= np.sum(w)  # Renormalize to sum to 1
    
    # Store records
    stumps.append(stump)
    alphas.append(alpha)
    errors.append(err)
    weights_history.append(w.copy())

# ==============================================================================
# 3. Two-Sided Visualization
# ==============================================================================
# Define 2D mesh grid for continuous decision boundary plotting
x_min, x_max = X[:, 0].min() - 0.7, X[:, 0].max() + 0.7
y_min, y_max = X[:, 1].min() - 0.7, X[:, 1].max() + 0.7
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 250),
                     np.linspace(y_min, y_max, 250))
grid_points = np.c_[xx.ravel(), yy.ravel()]

# Build layout: Left side (2x2 stumps), Right side (Dataset + Final Model)
fig = plt.figure(figsize=(18, 10), constrained_layout=True)
gs = gridspec.GridSpec(2, 4, figure=fig, width_ratios=[1, 1, 1.3, 1.3])

base_marker_size = 45
pos_mask = (y == 1)
neg_mask = (y == -1)

# ------------------------------------------------------------------------------
# Left Side: Individual Stumps (Rounds 1 to 4)
# ------------------------------------------------------------------------------
stump_coords = [(0, 0), (0, 1), (1, 0), (1, 1)]

for t in range(N_ROUNDS):
    row, col = stump_coords[t]
    ax = fig.add_subplot(gs[row, col])
    
    # Weak learner decision boundary
    Z_stump = stumps[t].predict(grid_points).reshape(xx.shape)
    ax.contourf(xx, yy, Z_stump, levels=[-1.5, 0, 1.5], alpha=0.25, cmap='bwr')
    ax.contour(xx, yy, Z_stump, levels=[0], colors='black', linewidths=1.5, linestyles='--')
    
    # Scale marker size proportional to sample weight at round t
    sizes = weights_history[t] * n_samples * base_marker_size
    ax.scatter(X[neg_mask, 0], X[neg_mask, 1], s=sizes[neg_mask],
               c='#d62728', marker='o', edgecolors='k', alpha=0.85, label='Class -1')
    ax.scatter(X[pos_mask, 0], X[pos_mask, 1], s=sizes[pos_mask],
               c='#1f77b4', marker='+', linewidths=2.5, alpha=0.9, label='Class +1')
    
    ax.set_title(f"Round {t + 1} (Stump {t + 1})", fontsize=12, fontweight='bold')
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_xlim(x_min, x_max)
    ax.set_ylim(y_min, y_max)
    ax.grid(True, linestyle=':', alpha=0.6)
    
    # Display error and alpha stats
    stats_text = f"Error ($\\epsilon$) = {errors[t]:.2f}\nAlpha ($\\alpha$) = {alphas[t]:.2f}"
    ax.text(0.04, 0.05, stats_text, transform=ax.transAxes, fontsize=10,
            verticalalignment='bottom', bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.8))

# ------------------------------------------------------------------------------
# Right Side: Top = Original Dataset
# ------------------------------------------------------------------------------
ax_data = fig.add_subplot(gs[0, 2:])
ax_data.scatter(X[neg_mask, 0], X[neg_mask, 1], s=base_marker_size,
                c='#d62728', marker='o', edgecolors='k', alpha=0.9, label='Class -1')
ax_data.scatter(X[pos_mask, 0], X[pos_mask, 1], s=base_marker_size * 1.5,
                c='#1f77b4', marker='+', linewidths=2.5, alpha=0.9, label='Class +1')

ax_data.set_title("Original Dataset (Equal Weights)", fontsize=13, fontweight='bold')
ax_data.set_xlabel("$x_1$")
ax_data.set_ylabel("$x_2$")
ax_data.set_xlim(x_min, x_max)
ax_data.set_ylim(y_min, y_max)
ax_data.grid(True, linestyle=':', alpha=0.6)
ax_data.legend(loc='upper right', framealpha=0.9)

# ------------------------------------------------------------------------------
# Right Side: Bottom = Final Combined Ensemble
# ------------------------------------------------------------------------------
ax_ensemble = fig.add_subplot(gs[1, 2:])

# Final hypothesis: H(x) = sign( sum_t alpha_t * h_t(x) )
ensemble_scores = np.zeros(grid_points.shape[0])
for alpha, stump in zip(alphas, stumps):
    ensemble_scores += alpha * stump.predict(grid_points)

Z_ensemble = np.sign(ensemble_scores).reshape(xx.shape)

# Ensemble training predictions and accuracy
train_scores = np.zeros(n_samples)
for alpha, stump in zip(alphas, stumps):
    train_scores += alpha * stump.predict(X)
final_preds = np.sign(train_scores)
acc = np.mean(final_preds == y) * 100

ax_ensemble.contourf(xx, yy, Z_ensemble, levels=[-1.5, 0, 1.5], alpha=0.3, cmap='bwr')
ax_ensemble.contour(xx, yy, Z_ensemble, levels=[0], colors='black', linewidths=2.2)

ax_ensemble.scatter(X[neg_mask, 0], X[neg_mask, 1], s=base_marker_size,
                    c='#d62728', marker='o', edgecolors='k', alpha=0.85, label='Class -1')
ax_ensemble.scatter(X[pos_mask, 0], X[pos_mask, 1], s=base_marker_size * 1.5,
                    c='#1f77b4', marker='+', linewidths=2.5, alpha=0.9, label='Class +1')

ax_ensemble.set_title("Final Combined Ensemble ($H(x)$)", fontsize=13, fontweight='bold')
ax_ensemble.set_xlabel("$x_1$")
ax_ensemble.set_ylabel("$x_2$")
ax_ensemble.set_xlim(x_min, x_max)
ax_ensemble.set_ylim(y_min, y_max)
ax_ensemble.grid(True, linestyle=':', alpha=0.6)
ax_ensemble.legend(loc='lower left', framealpha=0.9)

# Formula annotation card
formula_text = f"$H(x) = \\mathrm{{sign}}\\left(\\sum_{{t=1}}^{{{N_ROUNDS}}} \\alpha_t h_t(x)\\right)$\nAccuracy = {acc:.1f}%"
ax_ensemble.text(0.96, 0.88, formula_text, transform=ax_ensemble.transAxes, fontsize=11,
                 horizontalalignment='right', verticalalignment='top',
                 bbox=dict(boxstyle='round,pad=0.4', facecolor='white', edgecolor='gray', alpha=0.9))

plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.inspection import DecisionBoundaryDisplay

# Set seed for reproducible classroom demos
np.random.seed(42)

# Generate a 2D dataset that cannot be separated by a single straight line
X, y = make_moons(n_samples=7, noise=0.25, random_state=42)
# Convert labels from {0, 1} to {-1, +1} so the AdaBoost math is crystal clear
y_signed = np.where(y == 0, -1, 1)

plt.figure(figsize=(7, 5))
plt.scatter(X[y_signed == 1, 0], X[y_signed == 1, 1], c="#1f77b4", edgecolors="k", s=80, label="Class +1 (Blue)")
plt.scatter(X[y_signed == -1, 0], X[y_signed == -1, 1], c="#ff7f0e", edgecolors="k", s=80, label="Class -1 (Orange)")
plt.title("Can a single horizontal or vertical line separate these classes?")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

````python
n_rounds = 4
n_samples = len(X)

# Initialize all sample weights equally: w_i = 1 / N
weights = np.ones(n_samples) / n_samples

stumps = []
alphas = []

# Setup plotting grid
fig, axes = plt.subplots(n_rounds, 2, figsize=(13, 4 * n_rounds))
xx, yy = np.meshgrid(
    np.linspace(X[:, 0].min() - 0.5, X[:, 0].max() + 0.5, 200),
    np.linspace(X[:, 1].min() - 0.5, X[:, 1].max() + 0.5, 200)
)
grid_points = np.c_[xx.ravel(), yy.ravel()]

for t in range(n_rounds):
    # 1. Fit a weak learner (Decision Stump: max_depth=1) using current weights
    stump = DecisionTreeClassifier(max_depth=1, random_state=42)
    stump.fit(X, y_signed, sample_weight=weights)
    predictions = stump.predict(X)
    
    # 2. Calculate weighted error (epsilon) and Amount of Say (alpha)
    misclassified = (predictions != y_signed)
    error = np.sum(weights[misclassified])
    alpha = 0.5 * np.log((1 - error) / max(error, 1e-10))
    
    stumps.append(stump)
    alphas.append(alpha)
    
    # --- LEFT PLOT: Current Stump & Current Sample Weights ---
    ax_left = axes[t, 0]
    Z_stump = stump.predict(grid_points).reshape(xx.shape)
    ax_left.contourf(xx, yy, Z_stump, alpha=0.25, cmap="coolwarm_r")
    
    # Scale dot sizes by their current weight so students see mistakes grow!
    dot_sizes = weights * n_samples * 70
    ax_left.scatter(X[y_signed == 1, 0], X[y_signed == 1, 1], c="#1f77b4", s=dot_sizes[y_signed == 1], edgecolors="k")
    ax_left.scatter(X[y_signed == -1, 0], X[y_signed == -1, 1], c="#ff7f0e", s=dot_sizes[y_signed == -1], edgecolors="k")
    
    # Circle the mistakes made in this round in red
    ax_left.scatter(
        X[misclassified, 0], X[misclassified, 1],
        facecolors="none", edgecolors="red", linewidths=2.5,
        s=dot_sizes[misclassified] + 120, label="Misclassified (Will grow next round!)"
    )
    ax_left.set_title(f"Round {t+1} Weak Learner | Error = {error:.2f} | Say (α) = {alpha:.2f}")
    ax_left.legend(loc="lower right", fontsize=9)
    
    # --- RIGHT PLOT: Combined Ensemble So Far ---
    ax_right = axes[t, 1]
    ensemble_scores = sum(a * s.predict(grid_points) for s, a in zip(stumps, alphas))
    Z_ensemble = np.sign(ensemble_scores).reshape(xx.shape)
    
    ax_right.contourf(xx, yy, Z_ensemble, alpha=0.25, cmap="coolwarm_r")
    ax_right.scatter(X[y_signed == 1, 0], X[y_signed == 1, 1], c="#1f77b4", s=70, edgecolors="k")
    ax_right.scatter(X[y_signed == -1, 0], X[y_signed == -1, 1], c="#ff7f0e", s=70, edgecolors="k")
    
    ens_preds = np.sign(sum(a * s.predict(X) for s, a in zip(stumps, alphas)))
    ens_acc = np.mean(ens_preds == y_signed) * 100
    ax_right.set_title(f"Combined Ensemble (Stumps 1 to {t+1}) | Accuracy: {ens_acc:.1f}%")
    
    # 3. Update and normalize weights for the NEXT round
    weights = weights * np.exp(-alpha * y_signed * predictions)
    weights = weights / np.sum(weights)

plt.tight_layout()
plt.show()
````

````python
errors = np.linspace(0.01, 0.99, 200)
alphas_curve = 0.5 * np.log((1 - errors) / errors)

plt.figure(figsize=(8, 4.5))
plt.plot(errors, alphas_curve, color="#2ca02c", linewidth=3)
plt.axhline(0, color="black", linestyle="--", alpha=0.6)
plt.axvline(0.5, color="red", linestyle="--", alpha=0.6, label="Coin Flip (Error = 0.5 → Say = 0)")

plt.annotate("Strong Stump\n(High positive vote)", xy=(0.1, 1.1), xytext=(0.18, 1.5),
             arrowprops=dict(arrowstyle="->", lw=1.5), fontsize=10, fontweight="bold")
plt.annotate("Worse than guessing!\n(Vote is inverted)", xy=(0.85, -0.85), xytext=(0.62, -1.6),
             arrowprops=dict(arrowstyle="->", lw=1.5), fontsize=10, fontweight="bold")

plt.title("How a Stump's Error Rate Determines Its Voting Power (α)")
plt.xlabel("Weighted Error Rate (ε)")
plt.ylabel("Amount of Say (α)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# =====================================================================
# 1. GENERATE MONOTONIC NON-LINEAR DATA (Quadratic Curve: y = 10*(x/6)^2)
#    (Allows 1-split stumps in AdaBoost.R2 to shift splits each round!)
# =====================================================================
np.random.seed(42)
X_full = np.sort(np.random.uniform(0, 6, 80)).reshape(-1, 1)
y_true_func = lambda x: 10.0 * ((x.ravel() / 6.0) ** 2)
y_full = y_true_func(X_full) + np.random.normal(0, 0.45, X_full.shape[0])

# Split into Train (for step-by-step demo) and Test (for bias-variance check)
X, X_test, y, y_test = train_test_split(X_full, y_full, test_size=0.3, random_state=42)
sort_idx = np.argsort(X.ravel())
X, y = X[sort_idx], y[sort_idx]

x_grid = np.linspace(0, 6, 400).reshape(-1, 1)
y_grid_true = y_true_func(x_grid)

# =====================================================================
# 2. STEP-BY-STEP ADABOOST.R2 WITH TRUE STUMPS (max_depth=1)
# =====================================================================
n_rounds = 20
n_samples = len(X)
weights = np.ones(n_samples) / n_samples  # Start with equal weights: 1/N

stumps = []
alphas = []

# Create a 5-row x 2-column dashboard (Rows 1-4: Step-by-Step, Row 5: Summary)
fig, axes = plt.subplots(n_rounds + 1, 2, figsize=(15, 4.2 * (n_rounds + 1)))

for t in range(n_rounds):
    # A. Fit a TRUE REGRESSION STUMP (max_depth=1 -> 1 split, 2 flat lines)
    stump = DecisionTreeRegressor(max_depth=1, random_state=42)
    stump.fit(X, y, sample_weight=weights)
    y_pred = stump.predict(X)
    y_grid_pred = stump.predict(x_grid)
    split_point = stump.tree_.threshold[0]  # Exact X value where the stump splits
    
    # B. AdaBoost.R2 Math: Squared Relative Error (L_i), Beta, and Say (Alpha)
    abs_errors = np.abs(y - y_pred)
    max_error = np.max(abs_errors)
    rel_errors = (abs_errors / max(max_error, 1e-10)) ** 2  # Squared loss L_i in [0, 1]
    weighted_error = np.sum(weights * rel_errors)           # Average weighted loss
    
    beta = weighted_error / max(1.0 - weighted_error, 1e-10)
    alpha = np.log(1.0 / max(beta, 1e-10))                  # Stump voting power
    
    stumps.append(stump)
    alphas.append(alpha)
    
    # --- LEFT PANEL: Current 1-Split Stump, Weights & Worst Residuals ---
    ax_left = axes[t, 0]
    ax_left.plot(x_grid, y_grid_true, "k--", alpha=0.35, label="True Curve")
    ax_left.plot(x_grid, y_grid_pred, color="#2ca02c", lw=3,
                 label=f"Round {t+1} Stump (Split at X={split_point:.2f})")
    ax_left.axvline(split_point, color="#2ca02c", linestyle=":", alpha=0.85, lw=2)
    
    # Scale scatter dot sizes by current sample weights
    dot_sizes = weights * n_samples * 65
    ax_left.scatter(X, y, s=dot_sizes, c="#1f77b4", alpha=0.75, edgecolors="k", label="Data (Size = Weight)")
    
    # Highlight top 25% worst errors with red vertical residual bars
    worst_mask = rel_errors >= np.percentile(rel_errors, 75)
    for xi, yi, ypi in zip(X[worst_mask], y[worst_mask], y_pred[worst_mask]):
        ax_left.plot([xi[0], xi[0]], [yi, ypi], color="red", lw=2, alpha=0.85)
    ax_left.scatter(X[worst_mask], y[worst_mask], facecolors="none", edgecolors="red",
                    s=dot_sizes[worst_mask] + 70, lw=2, label="Large Error (Grows next round!)")
    
    ax_left.set_title(f"Round {t+1} Stump (max_depth=1) | Split X={split_point:.2f} | Loss={weighted_error:.2f} | Say (α)={alpha:.2f}")
    ax_left.set_ylim(-1.0, 11.5)
    ax_left.legend(loc="upper left", fontsize=8.5, ncol=2)
    ax_left.grid(alpha=0.3)
    
    # --- RIGHT PANEL: Combined Ensemble of Stumps (Rounds 1 to t) ---
    ax_right = axes[t, 1]
    ens_grid_pred = np.average([s.predict(x_grid) for s in stumps], axis=0, weights=alphas)
    ens_train_pred = np.average([s.predict(X) for s in stumps], axis=0, weights=alphas)
    
    mse_so_far = mean_squared_error(y, ens_train_pred)
    r2_so_far = r2_score(y, ens_train_pred)
    
    ax_right.plot(x_grid, y_grid_true, "k--", alpha=0.35, label="True Curve")
    ax_right.plot(x_grid, ens_grid_pred, color="#d62728", lw=2.8, label=f"Ensemble (Stumps 1..{t+1})")
    
    # Draw faint vertical lines for all stump splits accumulated so far
    for prev_s in stumps:
        ax_right.axvline(prev_s.tree_.threshold[0], color="#d62728", linestyle=":", alpha=0.4, lw=1.5)
        
    ax_right.scatter(X, y, s=55, c="#1f77b4", alpha=0.65, edgecolors="k")
    ax_right.set_title(f"Combined Ensemble (1 to {t+1} Stumps) | MSE: {mse_so_far:.2f} | R²: {r2_so_far:.2f}")
    ax_right.set_ylim(-1.0, 11.5)
    ax_right.legend(loc="upper left", fontsize=9)
    ax_right.grid(alpha=0.3)
    
    # C. Update Sample Weights for Next Round: w_i <- w_i * beta^(1 - L_i)
    weights = weights * (beta ** (1.0 - rel_errors))
    weights = weights / np.sum(weights)

# =====================================================================
# 3. ROW 5: SKLEARN ADABOOSTREGRESSOR WITH STUMPS (max_depth=1)
# =====================================================================
# Bottom-Left: Compare 1 vs. 5 vs. 50 Stumps using sklearn's AdaBoostRegressor
ax_bl = axes[n_rounds, 0]
ax_bl.plot(x_grid, y_grid_true, "k--", alpha=0.35, label="True Curve")
ax_bl.scatter(X, y, s=45, c="#1f77b4", alpha=0.4, edgecolors="k", label="Train Data")

for n_est, col, ls in [(1, "#2ca02c", ":"), (5, "#ff7f0e", "-."), (50, "#d62728", "-")]:
    model = AdaBoostRegressor(
        estimator=DecisionTreeRegressor(max_depth=1),  # Strictly Stumps!
        n_estimators=n_est, learning_rate=0.5, loss="square", random_state=42
    )
    model.fit(X, y)
    ax_bl.plot(x_grid, model.predict(x_grid), color=col, lw=2.5, linestyle=ls,
               label=f"{len(model.estimators_)} Stump(s)")

ax_bl.set_title("sklearn AdaBoostRegressor (max_depth=1): 1 vs. 5 vs. 50 Stumps")
ax_bl.set_xlabel("Feature X")
ax_bl.set_ylabel("Target y")
ax_bl.set_ylim(-1.0, 11.5)
ax_bl.legend(loc="upper left", fontsize=9, ncol=2)
ax_bl.grid(alpha=0.3)

# Bottom-Right: Train vs. Test Error Across Boosting Rounds
ax_br = axes[n_rounds, 1]
ada_full = AdaBoostRegressor(
    estimator=DecisionTreeRegressor(max_depth=1),      # Strictly Stumps!
    n_estimators=50, learning_rate=0.5, loss="square", random_state=42
)
ada_full.fit(X, y)

train_mse = [mean_squared_error(y, pred) for pred in ada_full.staged_predict(X)]
test_mse = [mean_squared_error(y_test, pred) for pred in ada_full.staged_predict(X_test)]

rounds_axis = range(1, len(train_mse) + 1)
best_round = int(np.argmin(test_mse)) + 1

ax_br.plot(rounds_axis, train_mse, label="Training MSE", color="#1f77b4", lw=2.2)
ax_br.plot(rounds_axis, test_mse, label="Testing MSE", color="#ff7f0e", lw=2.2)
ax_br.axvline(best_round, color="green", linestyle=":", lw=2,
              label=f"Lowest Test MSE (Round {best_round}: {min(test_mse):.2f})")
ax_br.set_title(f"Bias Reduction Across {len(train_mse)} Stumps (max_depth=1)")
ax_br.set_xlabel("Number of Boosting Stumps (n_estimators)")
ax_br.set_ylabel("Mean Squared Error (Lower is Better)")
ax_br.legend(loc="upper right", fontsize=9)
ax_br.grid(alpha=0.3)

plt.tight_layout()
plt.show()
````
