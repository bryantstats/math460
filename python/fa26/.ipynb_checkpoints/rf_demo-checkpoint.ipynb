```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.ensemble import RandomForestRegressor

# 1. Load class data
class_df = pd.DataFrame({
    'sleep': [5, 8, 6, 7, 4, 9, 6.5, 7.5, 5.5, 8, 6, 5],
    'caffeine': [3, 1, 2, 0, 4, 0, 2, 1, 3, 0, 2, 1],
    'screen_time': [6, 3, 5, 4, 7, 2, 4, 3, 6, 2, 5, 4],
    'commute': [15, 45, 10, 30, 20, 5, 25, 40, 12, 50, 18, 35],
    'alertness': [3, 8, 5, 7, 2, 9, 6, 8, 4, 9, 5, 7]
})

X = class_df[['sleep', 'caffeine', 'screen_time', 'commute']]
y = class_df['alertness']

# 2. Fit models
tree = DecisionTreeRegressor(max_depth=None, random_state=42).fit(X, y)
rf = RandomForestRegressor(n_estimators=100, max_features=2, random_state=42).fit(X, y)

# 3. Plot Single Tree vs. RF Feature Importances
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

plot_tree(tree, feature_names=X.columns, filled=True, ax=axes[0], fontsize=8)
axes[0].set_title("Single Unpruned Tree (Fits Every Student Noise Point)")

axes[1].barh(X.columns, rf.feature_importances_)
axes[1].set_title("Random Forest: Aggregated Feature Importance")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor, plot_tree
from sklearn.ensemble import RandomForestRegressor

# 1. Load class data
class_df = pd.DataFrame({
    'sleep': [5, 8, 6, 7, 4, 9, 6.5, 7.5, 5.5, 8, 6, 7],
    'caffeine': [3, 1, 2, 0, 4, 0, 2, 1, 3, 0, 2, 1],
    'screen_time': [6, 3, 5, 4, 7, 2, 4, 3, 6, 2, 5, 4],
    'commute': [15, 45, 10, 30, 20, 5, 25, 40, 12, 50, 18, 35],
    'alertness': [3, 8, 5, 7, 2, 9, 6, 8, 4, 9, 5, 7]
})

X = class_df[['sleep', 'caffeine', 'screen_time', 'commute']]
y = class_df['alertness']

# 2. Fit models
tree = DecisionTreeRegressor(max_depth=None, random_state=42).fit(X, y)
rf = RandomForestRegressor(n_estimators=100, max_features=2, random_state=42).fit(X, y)

# 3. Plot Single Tree vs. RF Feature Importances
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

plot_tree(tree, feature_names=X.columns, filled=True, ax=axes[0], fontsize=8)
axes[0].set_title("Single Unpruned Tree (Fits Every Student Noise Point)")

axes[1].barh(X.columns, rf.feature_importances_)
axes[1].set_title("Random Forest: Aggregated Feature Importance")
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Dataset 1: Original
X1 = class_df[['sleep', 'caffeine', 'screen_time', 'commute']]
y1 = class_df['alertness']

# Dataset 2: Modified (sleep changed from 7 to 5 at row index 3)
X2 = X1.copy()
X2.loc[3, 'sleep'] = 5
y2 = y1.copy()

# Fit Models
tree1 = DecisionTreeRegressor(max_depth=3, random_state=42).fit(X1, y1)
tree2 = DecisionTreeRegressor(max_depth=3, random_state=42).fit(X2, y2)

rf1 = RandomForestRegressor(n_estimators=200, random_state=42).fit(X1, y1)
rf2 = RandomForestRegressor(n_estimators=200, random_state=42).fit(X2, y2)

# Create grid across sleep values
sleep_grid = np.linspace(4, 10, 100).reshape(-1, 1)
grid_df = pd.DataFrame({
    'sleep': sleep_grid.flatten(),
    'caffeine': X1['caffeine'].median(),
    'screen_time': X1['screen_time'].median(),
    'commute': X1['commute'].median()
})

# Generate Predictions
pred_tree1 = tree1.predict(grid_df)
pred_tree2 = tree2.predict(grid_df)

pred_rf1 = rf1.predict(grid_df)
pred_rf2 = rf2.predict(grid_df)

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Single Tree
axes[0].plot(grid_df['sleep'], pred_tree1, 'r--', label='Dataset 1 (Original)', linewidth=2)
axes[0].plot(grid_df['sleep'], pred_tree2, 'r-', label='Dataset 2 (Sleep 7->5)', linewidth=2)
axes[0].set_title("Single Tree: Step Boundaries Shift")
axes[0].set_xlabel("Sleep (Hours)")
axes[0].set_ylabel("Predicted Alertness")
axes[0].legend()

# Random Forest
axes[1].plot(grid_df['sleep'], pred_rf1, 'b--', label='Dataset 1 (Original)', linewidth=2)
axes[1].plot(grid_df['sleep'], pred_rf2, 'b-', label='Dataset 2 (Sleep 7->5)', linewidth=2)
axes[1].set_title("Random Forest: Lines Overlap (Stable)")
axes[1].set_xlabel("Sleep (Hours)")
axes[1].set_ylabel("Predicted Alertness")
axes[1].legend()

plt.tight_layout()
plt.show()
```
